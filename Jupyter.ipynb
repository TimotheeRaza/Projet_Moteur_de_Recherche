```python
from Corpus import Corpus
from SearchEngine import SearchEngine

# Chargement du corpus depuis le fichier CSV
corpus = Corpus("Mon Corpus Projet")
corpus.load_from_csv("corpus_arxiv_reddit.csv")

print(f"Corpus chargé avec succès : {len(corpus.documents)} documents trouvés.")
```

```python
# Cellule suivante : Initialisation du SearchEngine
moteur = SearchEngine(corpus)
print("Moteur de recherche initialisé et matrices TF-IDF calculées avec succès !")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Création des composants graphiques
titre_label = widgets.HTML(value="<h2 style='color: #2b547e;'>🔍 Moteur de recherche M1</h2>")

input_mots_cles = widgets.Text(
    value='',
    placeholder='Saisissez vos mots-clés ici...',
    description='Mots clés :',
    disabled=False
)

slider_nb_doc = widgets.IntSlider(
    value=5,
    min=1,
    max=20,
    step=1,
    description='Nb résultats :',
    continuous_update=False
)

bouton_rechercher = widgets.Button(
    description='Rechercher',
    button_style='primary',
    icon='search'
)

output_resultats = widgets.Output()

# Fonction appelée au clic sur le bouton
def clique_bouton(b):
    requete = input_mots_cles.value
    n_docs = slider_nb_doc.value
    
    with output_resultats:
        clear_output()
        if not requete.strip():
            print("⚠️ Veuillez saisir au moins un mot-clé.")
            return
            
        print(f"Recherche en cours pour : '{requete}' (Top {n_docs})...\n")
        resultats_df = moteur.search(requete, n=n_docs)
        
        if resultats_df is None or resultats_df.empty:
            print("Aucun document pertinent trouvé.")
        else:
            print(f"-> {len(resultats_df)} document(s) trouvé(s) :\n")
            display(resultats_df)

bouton_rechercher.on_click(clique_bouton)
print("Composants de l'interface prêts !")
```

```python
import ipywidgets as widgets
from IPython.display import display, Markdown

filtre_source = widgets.Dropdown(
    options=['Tous', 'reddit', 'arxiv'],
    value='Tous',
    description='Source :',
    style={'description_width': 'initial'}
)

def recherche_avec_filtre(requete, source_choisie):
    resultats = moteur.search(requete) 
    
    if hasattr(resultats, "empty") and resultats.empty:
        display(Markdown(f"### 🔍 Aucun résultat trouvé pour : '{requete}'"))
        return

    if source_choisie != 'Tous' and hasattr(resultats, "__getitem__"):
        try:
            resultats = resultats[resultats['origine'] == source_choisie]
        except Exception:
            pass
    
    display(Markdown(f"### 🔍 Résultats pour : '{requete}' (Source : {source_choisie})"))
    display(resultats) # Affiche directement le DataFrame filtré

widgets.interact(
    recherche_avec_filtre, 
    requete="climb", 
    source_choisie=filtre_source
)
```
